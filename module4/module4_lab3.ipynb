```python
import pandas as pd

df = pd.read_csv("truck.csv")

df.head()
```

```python
r = df["X"].corr(df["Y"])

print("Pearson correlation:", r)
```

```python
ax = df.plot.scatter(
    x="X",
    y="Y",
    title="Truck List Price and Best Price Have a Strong Positive Relationship")

ax.set_xlabel("List Price ($1,000s)")
ax.set_ylabel("Best Price ($1,000s)")
```

```python
import numpy as np

slope, intercept = np.polyfit(df["X"], df["Y"], 1)

list_price = 25.2
predicted_price = slope * list_price + intercept

print("Predicted best price:", predicted_price)
```

I am highly confident in the prediction because the correlation of about 0.999 shows that list price and best price are very closely related. However, the data does not go beyond a list price of $22,400, so  predicting  the  best price  for a $25,200 truck is not a perfect science.
