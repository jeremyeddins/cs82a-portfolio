```python
import pandas as pd

df = pd.read_csv("allbacks.csv")

df.head()
```

```python
sorted_covers = df["cover"].sort_values().tolist()

print(sorted_covers)
print("Median cover:", sorted_covers[len(sorted_covers) // 2])
```

The median cover type is pb, but the median alone does not tell us much about which cover type is preferable because it only identifies the middle value of the ordered categories.

```python
mean_weight = df["weight"].mean()

print("Mean weight:", mean_weight)
```

The mean weight is 706.67 grams across all 15 books.

```python
cover_means = df.groupby("cover")["weight"].mean()

print(cover_means)
```

Hardbacks are about 168 grams heavier than paperbacks on average. The overall mean is closer to the paperback mean, which may partly reflect there being more paperbacks in the sample.
