# Temperature Categories
Cold: below 15°C

Warm: 15°C through 25°C, inclusive

Hot: above 25°C

```python
temperatures = [10, 16, 23, 28, 14, 22, 30]

for temp in temperatures:
    if temp < 15:
        label = "Cold"
    elif temp <= 25:
        label = "Warm"
    else:
        label = "Hot"

    print(temp, "°C:", label)
```
