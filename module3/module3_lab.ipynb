```python
import pandas as pd

df = pd.read_csv("messy_sales.csv")

print(df.shape)
display(df.head(10))
print(df.dtypes)
```

Three problems I see:

1. The date formats are all over the place.
2. Some products have NaN listed for their price, which will mess up the data.
3. One of the ZIP codes is only four digits.

```python
df.isna().sum()
```

```python
fill_value = df["price"].median()
print(fill_value)

df["price"] = df["price"].fillna(fill_value)

df.isna().sum()
```

```python
print(df.duplicated().sum())

df = df.drop_duplicates()

print(df.shape)
```

```python
df["zip"] = df["zip"].astype(str).str.zfill(5)

df["zip"].head(10)
```

```python
df["date"] = pd.to_datetime(df["date"], format="mixed")

df.dtypes
```

```python
df[df["qty"] < 0]
```

```python
returns = df[df["qty"] < 0]
returns
```

The negative values represent returned goods, so that is how I will treat them. Removing or changing them could have a negative impact on my analysis.

```python
df.to_csv("sales_clean.csv", index=False)
```
