```python
def multiply_for(a, b):
    result = 0

    for i in range(b):
        result = result + a

    return result

print(multiply_for(7, 6))
```

```python
def multiply_while(a, b):
    result = 0
    counter = 0

    while counter < b:
        result = result + a
        counter = counter + 1

    return result

print(multiply_while(7, 6))
```

```python
print("7 x 6:")
print("For loop:", multiply_for(7, 6))
print("While loop:", multiply_while(7, 6))

print("5 x 4:")
print("For loop:", multiply_for(5, 4))
print("While loop:", multiply_while(5, 4))

print("9 x 3:")
print("For loop:", multiply_for(9, 3))
print("While loop:", multiply_while(9, 3))
```

The while loop starts the result at 0 and repeatedly adds 7 while the counter is less than 6. The counter starts at 0 and increases by 1 each time the loop runs. Once the counter reaches 6, the condition is no longer true, so the loop stops and returns the final result of 42.
